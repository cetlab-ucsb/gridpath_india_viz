```python
path_to_scen = '/Users/Guille/Desktop/india_power/scenarios/cost/'

scens_ = [
    'VREhigh_SThigh_CONVhigh_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VREhigh_SThigh_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VREmid_STmid_CONVhigh_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VREmid_STmid_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VRElow_STlow_CONVhigh_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VRElow_STlow_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
]

for scen in scens_:
    print(scen)
    
    df = pd.read_csv(path_to_scen + scen + '/inputs/new_build_generator_vintage_costs.tab', sep="\t")
    
    df_coal = df.loc[df.project.str.contains('Coal') & (df['vintage'] == 2050)].reset_index(drop = True)
    print(df_coal['annualized_real_cost_per_mw_yr'].unique())
    
    df_solar = df.loc[df.project.str.contains('solarPV_single') & (df['vintage'] == 2050)].reset_index(drop = True)
    print(df_solar['annualized_real_cost_per_mw_yr'].unique())
    
    df_wind = df.loc[df.project.str.contains('wind') & (df['vintage'] == 2050)].reset_index(drop = True)
    print(df_wind['annualized_real_cost_per_mw_yr'].unique())
```

```python
path_to_scen = '/Users/Guille/Desktop/india_power/scenarios/cost/'

scens_ = [
    'VREhigh_SThigh_CONVhigh_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VREhigh_SThigh_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VREmid_STmid_CONVhigh_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VREmid_STmid_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VRElow_STlow_CONVhigh_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
    'VRElow_STlow_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid',
]

for scen in scens_:
    print(scen)
        
    df = pd.read_csv(path_to_scen + scen + '/results/project_timepoint.csv', low_memory=False)
    df_sc = df.loc[((df['technology'] == 'Supercritical_Coal') 
                    & (df['period'] == 2050))].reset_index(drop = True)
    
    df_sc['weighted_power_mw'] = (df_sc['power_mw'] 
                                  * df_sc['timepoint_weight'] 
                                  * df_sc['number_of_hours_in_timepoint'])
    
    print(df_sc.shape, df_sc['weighted_power_mw'].sum())

    df_scs = df.loc[((df['technology'] == 'Subcritical_Coal_Small') 
                    & (df['period'] == 2050))].reset_index(drop = True)
    
    df_scs['weighted_power_mw'] = (df_scs['power_mw'] 
                                   * df_scs['timepoint_weight'] 
                                   * df_scs['number_of_hours_in_timepoint'])
    
    print(df_scs.shape, df_scs['weighted_power_mw'].sum())

    df_scl = df.loc[((df['technology'] == 'Subcritical_Coal_Large') 
                    & (df['period'] == 2050))].reset_index(drop = True)
    
    df_scl['weighted_power_mw'] = (df_scl['power_mw'] 
                                   * df_scl['timepoint_weight'] 
                                   * df_scl['number_of_hours_in_timepoint'])
    
    print(df_scl.shape, df_scl['weighted_power_mw'].sum())
    print(df_scl['weighted_power_mw'].sum() + df_scs['weighted_power_mw'].sum() + df_sc['weighted_power_mw'].sum())
```
