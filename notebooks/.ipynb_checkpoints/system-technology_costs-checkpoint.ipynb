```python
import os

import pandas as pd
import numpy as np
import geopandas as gpd
import matplotlib.pyplot as plt
import loading as ld
import processing as prc
import visualization as viz

from matplotlib.transforms import ScaledTranslation
from matplotlib.lines import Line2D

path_to_images = '/Users/Guille/Desktop/india_power/images'
path_to_csvs = '/Users/Guille/Desktop/india_power/gridpath_india_viz/csvs'
path_to_output = '/Users/Guille/Desktop/india_power/output_csvs'
```

```python
tech_labels_ = pd.read_csv(path_to_csvs + '/technology_labels.csv') 
```

# Main Figure

Figure: Fixed and variable costs by technology
* Technology cost scenarios
* Make a version for other technology groups

```python
costs_ = {}
costs_['scen_labels'] = pd.read_csv(path_to_csvs + '/cost-scenario_labels.csv') 

cost_, demand_ = ld._load_cost(costs_['scen_labels'])

cost_ = prc._group_technology(cost_, tech_labels_)

costs_['tech_cost'], costs_['demand'] = prc._group_cost_by_technology(cost_, demand_, costs_['scen_labels'])

# Add fixed cost of specified transmission capacity exogenously
costs_['tech_cost'].loc[costs_['tech_cost']['technology'] == 'Transmission Losses', 
                        'fixed_cost'] += 5.79e9

prc._save_technology_cost_csv(costs_, path_to_output, 
                              file_name = 'cost-technology_costs.csv')
```

```python
fig, _ax = plt.subplot_mosaic([['A', 'B']], 
                              layout = 'constrained', 
                              figsize = (20, 8.5), 
                              gridspec_kw = {'width_ratios': [1, 1]})

for label, ax in _ax.items():
    ax.text(0.0, 1.0, label, 
            transform = (ax.transAxes + ScaledTranslation(-20/72, +7/72, fig.dpi_scale_trans)),
            fontsize = 20, 
            weight = "bold",
            va = 'bottom',
            family = 'Arial')

# Enter a title of your choice between the quotation marks ('') under title = ''.
viz._plot_total_fixed_and_variable_cost(_ax['A'], costs_['tech_cost'], costs_['scen_labels'], tech_labels_, 
                                        ylabel = r'Total System Costs (Billion USD)',
                                        yplabel = r'(Lakh Crores INR)',
                                        y_lim_max = 400,
                                        y_grid_inc = 50,
                                        legend = False)
    
# Enter a title of your choice between the quotation marks ('') under title = ''.
viz._plot_average_fixed_and_variable_cost(_ax['B'], costs_['tech_cost'], costs_['demand'], costs_['scen_labels'], tech_labels_, 
                                          ylabel = r'Average System Costs (USD per MWh)',
                                          yplabel = r'(INR per kWh)',
                                          y_lim_max = 70,
                                          y_grid_inc = 10,
                                          legend = True)

line1    = Line2D([], [], color = 'None', solid_capstyle ='round')
handles_ = [line1, line1, line1, line1, line1, line1]
labels_  = list(costs_['scen_labels']['label'].to_numpy())

leg = _ax['A'].legend(handles_, labels_, 
                      loc = (0., -.4125),
                      title = 'Scenarios - Technology Cost (from left to right)',
                      frameon = False,
                      labelspacing = 0.175,
                      title_fontsize = 16,
                      ncol = 1,
                      handletextpad = -1.,
                      prop = {'size': 14})

leg._legend_box.align = "left"

plt.savefig(path_to_images + '/cost-technology_costs.pdf', 
            bbox_inches = 'tight')

plt.savefig(path_to_images + '/cost-technology_costs.png', 
            bbox_inches = 'tight', 
            dpi = 600)

plt.show()
```
