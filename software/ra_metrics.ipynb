```python
import os, string

import pandas as pd
import numpy as np
import itertools

dummy_ra_system_summary = pd.read_csv('../csvs/dummy_ra_system_summary.csv')
```

```python
# Define identifying and numeric columns
identifying_cols = ['weather_iteration', 'hydro_iteration', 'availability_iteration']
numeric_cols_to_sum = ['total_power_mw', 'net_imports_mw', 'overgeneration_mw', 'static_load_mw', 'unserved_energy_mw']

# Group by identifying columns and timepoint, then sum numeric columns
india_summary_grouped = dummy_ra_system_summary.groupby(
    identifying_cols + ['timepoint']
)[numeric_cols_to_sum].sum().reset_index()

# Assign 'India' to the 'load_zone' column
india_summary_grouped['load_zone'] = 'India'

# Reorder columns (optional, but good practice)
final_cols_order = identifying_cols + ['load_zone', 'timepoint'] + numeric_cols_to_sum
india_summary = india_summary_grouped[final_cols_order]

# Save the aggregated summary to a CSV file
output_path = '../csvs/india_summary_aggregated.csv'
india_summary.to_csv(output_path, index=False)
print(f'Aggregated summary saved to {output_path}')
```
