```python
df_ = pd.read_csv(path_to_data + 'studies_bess_capacity.csv')
df_ = df_.dropna()

fig, _ax = plt.subplot_mosaic([['A', 'B']], 
                              layout      = 'constrained', 
                              figsize     = (10.75, 6), 
                              gridspec_kw = {'width_ratios': [1, 1]})

for label, ax in _ax.items():
    ax.text(0.0, 1.0, label, transform = (ax.transAxes + ScaledTranslation(-20/72, +7/72, fig.dpi_scale_trans)),
                             fontsize  = 22, 
                             weight    = "bold",
                             va        = 'bottom',
                             family    = 'Arial')

def _plot_studies_bess_comparison(_ax, df_, column,
                                  ylabel = 'Installation Rate (GW/yr)', 
                                  units = 1000):
    
    for study in df_['study'].unique():
        df_p_ = df_.loc[df_['study'] == study].reset_index(drop = True)
        x_ = df_p_['period'].to_numpy().astype(int)
        y_ = df_p_[column].to_numpy()
    
        _ax.plot(x_, y_/units, 
                 label = study, 
                 color = df_p_['color'].to_numpy()[0],
                 lw    = df_p_['lw'].to_numpy()[0],
                 ls    = df_p_['ls'].to_numpy()[0])
    
    _ax.set_ylabel(ylabel, size = 18)
    _ax.tick_params(labelsize = 16)
    _ax.yaxis.set_major_formatter(mpl.ticker.StrMethodFormatter('{x:,.0f}'))

    x_ = [2030, 2040, 2050]
    _ax.set_xticks(x_, x_)
    _ax.set_ylim(0,)
    #_ax.set_xlim(x_[0], x_[-1])

_plot_studies_bess_comparison(_ax['A'], df_, 
                              column = 'capacity_mw',
                              ylabel = 'Battery Power (GW)')
    

_plot_studies_bess_comparison(_ax['B'], df_, 
                              column = 'capacity_mwh',
                              ylabel = 'Battery Energy (GWh)')
    

_ax['B'].legend(frameon        = False, 
                #title          = "Study", 
                fontsize       = 14, 
                title_fontsize = 16)

plt.savefig(path_to_images + f'/SI-battery_studies_comparison.png', 
            bbox_inches = 'tight', 
            dpi         = 600)

plt.show()
```

```python
# Grab data from databases for plotting new and existing capacity
def _load_capacity(scen_labels_):

    # Load project capacity table and process them from database
    def __load_new_and_existing_csv(df_, scenario):

        df_p_  = df_[['period', 
                      'technology',
                      'load_zone', 
                      'capacity_mw']].copy()

        df_p_ = df_p_.groupby(['period', 
                               'technology', 
                               'load_zone']).agg('sum').reset_index(drop = False)

        df_p_['scenario'] = scenario

        return df_p_
    
    dfs_ = []
    # Open connection: open database and grab meta-data
    for scen, path in zip(scen_labels_['scenario'],  scen_labels_['path']):
        print(scen, path)
        dir_name  = r'{}/{}'.format(path, scen)
        capacity_ = pd.read_csv(dir_name + r'/results/project_period.csv', low_memory = False)
        df_       = __load_new_and_existing_csv(capacity_, scen)
        
        # Load specified capacity from csv files
        dfs_.append(df_)

    return pd.concat(dfs_, axis = 0).reset_index(drop = True)

# Grab data from databases for plotting energy dispatch and clean energy targets
def _load_dispatch(scen_labels_):

    # Load energy dispatch table and process data from database
    def __load_dispatch_from_csv(df_, scenario):

        df_['dispatch_mw']  = df_['power_mw'] * df_['timepoint_weight'] * df_['number_of_hours_in_timepoint']
        df_['potential_mw'] = df_['capacity_mw'] * df_['timepoint_weight'] * df_['number_of_hours_in_timepoint']

        df_ = df_[['period', 
                   'timepoint',
                   'technology', 
                   'load_zone', 
                   'dispatch_mw', 
                   'potential_mw']]

        df_ = df_.groupby(['period', 
                           'timepoint',
                           'technology', 
                           'load_zone']).agg({'dispatch_mw': 'sum',
                                              'potential_mw': 'sum'}).reset_index(drop = False)
         
        df_['scenario'] = scenario
                
        return df_

    dfs_ = []
    # Open connection: open database and grab meta-data
    for scen, path in zip(scen_labels_['scenario'], scen_labels_['path']):
        print(scen, path)
        
        dir_name  = r'{}/{}'.format(path, scen)
        dispatch_ = pd.read_csv(dir_name + f'/results/project_timepoint.csv', low_memory = False)
        
        dfs_ += [__load_dispatch_from_csv(dispatch_, scen)]

    return pd.concat(dfs_, axis = 0).reset_index(drop = True)

def _group_capacity_technologies(capacity_, tech_labels_):

    for group in tech_labels_['group'].unique():
        capacity_.loc[capacity_['technology'].isin(tech_labels_.loc[tech_labels_['group'] == group, 'technology']), 'technology'] = group
        
    capacity_ = capacity_.groupby(['period', 
                                   'technology', 
                                   'load_zone', 
                                   'scenario']).agg('sum')

    return capacity_.reset_index(drop = False)

def _group_dispatch_technologies(dispatch_, tech_labels_):

    for group in tech_labels_['group'].unique():
        dispatch_.loc[dispatch_['technology'].isin(tech_labels_.loc[tech_labels_['group'] == group, 'technology']), 'technology'] = group
        
    dispatch_ = dispatch_.groupby(['period', 
                                   'timepoint',
                                   'technology', 
                                   'load_zone', 
                                   'scenario']).agg('sum')

    return dispatch_.reset_index(drop = False)
```

```python
def _installation_rate(x_, y_, Ro):

    # build continuous rate with piecewise constant acceleration
    t_ = np.linspace(x_[0], x_[-1], x_[-1] - x_[0] + 1)
    
    r_ = np.array([Ro])
    z_ = np.array([y_[0]])
    
    dx_ = np.diff(x_)
    dy_ = np.diff(y_)

    # loop over periods
    for i in range(dx_.shape[0]):
        T = dx_[i] * (dx_[i] + 1) / 2
        a = (dy_[i] - dx_[i]*r_[-1])/T
        # loop over years in a period
        for t in range(dx_[i]):
            r_ = np.append(r_, r_[-1] + a)
            z_ = np.append(z_, z_[-1] + r_[-1] )
            
    return r_, z_, t_

dperiod_ = period_[1:-1]
r_solar_ = np.diff(df_solar_)[1:]
r_wind_  = np.diff(df_wind_)[1:]

y_solar_1_[0] = df_solar_[-1]
y_solar_2_[0] = df_solar_[-1]

y_wind_1_[0] = df_wind_[-1]
y_wind_2_[0] = df_wind_[-1]

x_1_[0] = period_[-1]
x_2_[0] = period_[-1]

r_solar_1_, y_solar_1_r_, x_1_r_ = _installation_rate(x_1_, y_solar_1_, Ro = r_solar_[-1])
r_solar_2_, y_solar_2_r_, x_2_r_ = _installation_rate(x_2_, y_solar_2_, Ro = r_solar_[-1])

r_wind_1_, y_wind_1_r_, x_1_r_ = _installation_rate(x_1_, y_wind_1_, Ro = r_wind_[-1])
r_wind_2_, y_wind_2_r_, x_2_r_ = _installation_rate(x_2_, y_wind_2_, Ro = r_wind_[-1])

fig, _ax = plt.subplots(figsize = (5, 6.5))

_ax.plot(period_, df_solar_/1000, 
         label = 'Solar (actual)', 
         lw = 3.5, 
         c = '#daac77')

_ax.plot(period_, 
         df_wind_/1000, 
         label = 'Wind (actual)', 
         lw = 3.5, 
         c = '#8dc0cd')

_ax.plot(x_2_r_, 
         y_solar_2_r_/1000, 
         label = '90% Carbon Target - REF',
         ls = '-',
         lw = 1.75, 
         c = 'gray')

_ax.plot(x_2_r_, y_solar_2_r_/1000, 
         lw = 1.75, 
         ls = '-',
         c = '#daac77')

_ax.plot(x_2_, y_solar_2_/1000, 
         lw = .25, 
         ls = '-',
         c = 'k')

_ax.plot(x_2_r_, y_wind_2_r_/1000, 
         lw = 1.75, 
         ls = '-',
         c = '#8dc0cd')

_ax.plot(x_2_, y_wind_2_/1000, 
         lw = .25, 
         ls = '-',
         c = 'k')

_ax.plot(x_1_r_, 
         y_solar_1_r_/1000, 
         label = 'No Carbon Target',
         ls = '--',
         lw = 1.75, 
         c = 'gray')

_ax.plot(x_1_r_, 
         y_solar_1_r_/1000, 
         ls = '--',
         lw = 1.75, 
         c = '#daac77')

_ax.plot(x_1_, y_solar_1_/1000, 
         lw = .25, 
         ls = '-',
         c = 'k')

_ax.plot(x_1_r_, 
         y_wind_1_r_/1000, 
         ls = '--',
         lw = 1.75, 
         c = '#8dc0cd')

_ax.plot(x_1_, y_wind_1_/1000, 
         lw = .25, 
         ls = '-',
         c = 'k')



#_ax.set_xticks(2010 + np.arange(4, type = int)*10, 2010 + np.arange(4, type = int)*10)
plt.legend(frameon = False, fontsize = 14)

_ax.set_ylabel('Installed Capacity (GW)', size = 18)
_ax.tick_params(labelsize = 16)
#_ax.yaxis.set_major_formatter(mpl.ticker.StrMethodFormatter('{x:,.0f}'))
_ax.set_ylim(0,)

plt.savefig(path_to_images + f'/SI-installed_capacity.pdf', 
            bbox_inches = 'tight', 
            dpi         = 600)

plt.savefig(path_to_images + f'/SI-installed_capacity.png', 
            bbox_inches = 'tight', 
            dpi         = 600)

plt.show()
```

```python
def _installation_rate(x_, y_, Ro):

    # build continuous rate with piecewise constant acceleration
    t_ = np.linspace(x_[0], x_[-1], x_[-1] - x_[0] + 1)
    
    r_ = np.array([Ro])
    z_ = np.array([y_[0]])
    
    dx_ = np.diff(x_)
    dy_ = np.diff(y_)

    # loop over periods
    for i in range(dx_.shape[0]):
        T = dx_[i] * (dx_[i] + 1) / 2
        a = (dy_[i] - dx_[i]*r_[-1])/T
        # loop over years in a period
        for t in range(dx_[i]):
            r_ = np.append(r_, r_[-1] + a)
            z_ = np.append(z_, z_[-1] + r_[-1] )
            
    return r_, z_, t_

# example breakpoints (year, rate)
x_ = np.array([2020, 2030, 2040, 2050])
y_ = np.array([5, 20, 75, 115]) # GW

# Ro is the initial installation rate (it can be any number)
R_, y_p_, x_p_ = _installation_rate(x_, y_, Ro = 1)

plt.figure()
plt.plot(x_, y_)
plt.plot(x_p_, y_p_)
plt.ylabel('Capacity (MW)')
plt.show()
```

```python
# Grab data from databases for plotting energy dispatch and clean energy targets
def _load_inertia(scen_labels_, gen_inertia_):

    # Load energy dispatch table and process data from database
    def __load_inertia_from_csv(dispatch_, system_load_, tx_, gen_inertia_):
    

        # Calculate total system load
        charge_ = dispatch_.loc[dispatch_['power_mw'] < 0].reset_index(drop = True)

        charge_ = charge_.groupby(['period', 
                                   'timepoint']).agg({'power_mw': 'sum'}).reset_index(drop = False)

        system_load_ = system_load_.groupby(['period', 
                                             'timepoint']).agg({'static_load_mw': 'sum'}).reset_index(drop = False)


        tx_ = tx_.groupby(['period', 
                           'timepoint']).agg({'transmission_losses_lz_to': 'sum',
                                              'transmission_losses_lz_from': 'sum'}).reset_index(drop = False)

        df_ = pd.merge(charge_, system_load_, 
                       on = ['period', 'timepoint'], how = "right") 
        df_ = pd.merge(df_, tx_, 
                       on = ['period', 'timepoint'], how = "right") 

        df_['power_mw'] = df_['power_mw'].fillna(0)
        
        df_['load_mw'] = (df_['static_load_mw']
                         + df_['transmission_losses_lz_to']
                         + df_['transmission_losses_lz_from']
                         - df_['power_mw'])
        
        # Calculate total system innertia
        discharge_ = dispatch_.loc[dispatch_['power_mw'] > 0].reset_index(drop = True)
        discharge_ = discharge_.loc[discharge_['technology'].isin(gen_inertia_['technology'].tolist())].reset_index(drop = True)
        discharge_ = discharge_.groupby(['period', 
                                         'timepoint',
                                         'operational_type',
                                         'technology']).agg({'capacity_mw': 'sum',
                                                             'committed_mw': 'sum',
                                                             'power_mw': 'sum'}).reset_index(drop = False)

        discharge_['interia_mws'] = 0.

        for optype in discharge_['operational_type'].unique():
            for technology in discharge_.loc[discharge_['operational_type'] == optype, 'technology'].unique():

                idx_1_ = discharge_['technology'] == technology
                idx_2_ = gen_inertia_['technology'] == technology

                H = gen_inertia_.loc[idx_2_, 'avg'].to_numpy()[0]

                if optype == 'gen_commit_cap':
                    discharge_.loc[idx_1_, 'interia_mws'] = discharge_.loc[idx_1_, 'committed_mw'] * H
                                                           
                if optype == 'gen_hydro_must_take':
                    discharge_.loc[idx_1_, 'interia_mws'] = discharge_.loc[idx_1_, 'capacity_mw'] * H

                if optype == 'gen_must_run':
                    discharge_.loc[idx_1_, 'interia_mws'] = discharge_.loc[idx_1_, 'capacity_mw'] * H

                if optype == 'gen_var_must_take':
                    discharge_.loc[idx_1_, 'interia_mws'] = discharge_.loc[idx_1_, 'capacity_mw'] * H

                if optype == 'stor':
                    discharge_.loc[idx_1_, 'interia_mws'] = discharge_.loc[idx_1_, 'power_mw'] * H

        discharge_ = discharge_.groupby(['period', 
                                         'timepoint']).agg({'interia_mws': 'sum'}).reset_index(drop = False)

        df_ = pd.merge(df_, discharge_, 
                       on = ['period', 'timepoint'], 
                       how = "right") 

        df_['load-to-intertia'] = 1./(df_['load_mw']/df_['interia_mws'])

        df_ = df_[['period', 
                   'timepoint', 
                   'load_mw', 
                   'interia_mws', 
                   'load-to-intertia']]

        return df_

    dfs_ = []
    # Open connection: open database and grab meta-data
    for scen, path in zip(scen_labels_['scenario'], scen_labels_['path']):
        print(scen, path)
        
        dir_name  = r'{}/{}'.format(path, scen)
        
        dispatch_    = pd.read_csv(dir_name + f'/results/project_timepoint.csv', low_memory = False)
        system_load_ = pd.read_csv(dir_name + f'/results/system_load_zone_timepoint.csv', low_memory = False)
        tx_          = pd.read_csv(dir_name + f'/results/transmission_timepoint.csv', low_memory = False)

        dfs_ += [__load_inertia_from_csv(dispatch_, system_load_, tx_, gen_inertia_)]

    return pd.concat(dfs_, axis = 0).reset_index(drop = True)

gen_inertia_        = pd.read_csv(path_to_inputs + r'/generators/generators_inertia.csv')
gen_inertia_['avg'] = (gen_inertia_['min'] + gen_inertia_['max'])/2
gen_inertia_        = gen_inertia_[['technology', 'avg']]

inertia_ = _load_inertia(scen_labels_, gen_inertia_)
print(inertia_.shape)
```

```python
# Plot land use (MHa) for different scenarios
def _plot_capacity_factors(ax, df_1_, df_2_, tech_labels_, scen, 
                           load_zone    = None,
                           ylabel       = r'Capcity Factor (%)',
                           legend_title = 'Resource'):
    
    if load_zone == None:
        df_1_ = df_1_.groupby(['period', 
                               'technology', 
                               'scenario']).agg({'power_mw': 'sum'}).reset_index(drop = False)
    
        df_2_ = df_2_.groupby(['period', 
                               'technology', 
                               'scenario']).agg({'capacity_mw': 'sum'}).reset_index(drop = False)
        
    df_1p_ = df_1_.loc[(df_1_['scenario'] == scen)].reset_index(drop = True)    
    df_2p_ = df_2_.loc[(df_2_['scenario'] == scen)].reset_index(drop = True)    
    
    for tech in df_1p_['technology'].unique():  
        df_1pp_ = df_1p_.loc[df_1p_['technology'] == tech].sort_values(by = ['period']).reset_index(drop = True)
        df_2pp_ = df_2p_.loc[df_2p_['technology'] == tech].sort_values(by = ['period']).reset_index(drop = True)
        color   = tech_labels_.loc[tech_labels_['group'] == tech, 'group_color'].to_numpy()[0]
        
        cf_ = 100.*(df_1pp_['power_mw']/(df_2pp_['capacity_mw']*24*24))

        ax.plot(df_2pp_['period'], cf_, 
                label = tech,
                color = color,
                alpha = 0.75, 
                lw    = 2.5)
    
    ax.set_ylim(-.25, 100)
    ax.xaxis.set_tick_params(labelsize = 18)
    ax.yaxis.set_tick_params(labelsize = 18)
    ax.set_ylabel(ylabel, fontsize = 20)


scen = 'VREmid_STmid_CONVmid_H2_RES_10PRM_CC_50RPS_90CAP_500GW_PIERmid-month'

fig, ax = plt.subplots(1, 1, figsize = (4, 5))

_plot_capacity_factors(ax, grouped_dispatch_, grouped_capacity_, tech_labels_, scen)

plt.show()
```

```python
# Grab data from databases for plotting energy dispatch and clean energy targets
def _availability(scen, path):

    # Load energy dispatch table and process data from database
    def __availbility_from_csv(df_):

        df_ = df_[['project', 
                   'period', 
                   'technology',
                   'availability_derate', 
                   'timepoint', 
                   'capacity_mw', 
                   'power_mw',
                   'timepoint_weight']]

        df_['power_mw'] = df_['power_mw']*df_['timepoint_weight']
                 
        return df_
        _
    print(scen, path)

    dir_name  = r'{}/{}'.format(path, scen)
    projects_ = pd.read_csv(dir_name + f'/results/project_timepoint.csv', 
                            low_memory = False)

    return __availbility_from_csv(projects_).reset_index(drop = True)

period = 2050
tech   = 'CT'

path   = '/Users/Guille/Desktop/india_power/scenarios/demand'
scen   = 'VREmid_STmid_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_ICEDhigh'

availability_ = _availability(scen, path)
idx_          = (availability_['technology'] == tech) & (availability_['period'] == period)
availability_ = availability_.loc[idx_].reset_index(drop = True)
```
