### Battery storage

Battery storage, which is limited to short-term (one day) balancing energy in this study, plays a major role in balancing the diurnal variability of VRE generation, especially solar. In the reference scenario, battery storage or energy storage system (ESS) capacity rises from near-zero in 2020 to 154GW/308GWh by 2030, 401GW/802GWh by 2040, and 650GW/3,230GWh in 2050. The high power capacity relative to energy capacity in 2030 and 2040 is driven by battery storage's contribution to the reliability constraint of primary reserve margin (PRM) to see the difference between the capacity deployment in the reference scenario with and without PRM). In 2050, the battery energy capacity increases substantially---8-12 times the energy capacity in 2030---highlighting the greater need for balancing services as the share of weather-dependent energy sources like wind and solar increase to over 90\%. The variation in battery energy capacities in 2050 is because of technology costs and competition with green hydrogen storage.

As the share of near-zero variable cost wind and solar resources increases and the share of fuel-based fossil resources decreases in India's electricity system, the share of variable costs also commensurately shrinks. Out of total system costs, both absolute and per unit, the share of variable costs decreases from 45.1% in 2020 to 32.3-35.7% in 2030, 17.8-18.1% in 2040, and 3-4.1% in 2050. 

```python
df_ = pd.read_csv(path_to_tables + '/cost-technology_costs.csv')
print(df_['scenario'].unique())

df_ = df_.groupby(['period', 'scenario']).agg({'variable_cost': 'sum',
                                               'fixed_cost': 'sum'}).reset_index(drop = False)

df_['total_costs']         = df_['fixed_cost'] + df_['variable_cost']
df_['variable_cost_share'] = 100.*df_['variable_cost']/df_['total_costs']
#print(df_)

df_p_ = df_.groupby(['period']).agg({'variable_cost_share': 'min'}).reset_index(drop = False)
print(df_p_.round(1))
df_p_ = df_.groupby(['period']).agg({'variable_cost_share': 'max'}).reset_index(drop = False)
print(df_p_.round(1))
```

Total variable costs of coal increases from 2020 to 2030 as coal generation increases but then decreases substantially in 2040 and 2050 as coal generation diminishes. The share of variable costs within just coal's total costs falls from 53.5-53.7\% in 2030 to 40.9-47.9\% in 2040 and 8.4-8.8\% in 2050, highlighting coal plants increasing role in providing capacity for reliability and decreasing energy generation because of unfavorable economics and climate targets.

```python
df_ = pd.read_csv(path_to_tables + '/cost-total_tx_transfers.csv')

df_['power_mw'] = df_['power_mw'].abs()/1e-6
df_ = df_.groupby(['period', 
                   'scenario']).agg({'power_mw': 'sum'}).reset_index(drop = False)

idx_ = df_['scenario'] == 'VRE & ESS (low) Coal (low)'
print(df_.loc[idx_, 'power_mw'].to_numpy()/df_.loc[idx_, 'power_mw'].to_numpy()[0])

idx_ = df_['scenario'] == 'VRE & ESS (mid) Coal (low) - REF'
print(df_.loc[idx_, 'power_mw'].to_numpy()/df_.loc[idx_, 'power_mw'].to_numpy()[0])

idx_ = df_['scenario'] == 'VRE & ESS (high) Coal (low)'
print(df_.loc[idx_, 'power_mw'].to_numpy()/df_.loc[idx_, 'power_mw'].to_numpy()[0])
```

Transfers are especially large coming out of Rajasthan as the state exports its high quality renewable generation to other states.
```python
df_ = pd.read_csv(path_to_tables + '/cost-tx_capacity.csv')

df_ = df_.groupby(['period', 'scenario']).agg({'spec_mw': 'sum',
                                               'new_mw': 'sum'}).reset_index(drop = False)

idx_ = df_['scenario'] == 'VRElow_STlow_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid'
print(df_.loc[idx_, 'new_mw']/df_.loc[idx_, 'spec_mw'])

idx_ = df_['scenario'] == 'VREmid_STmid_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid'
print(df_.loc[idx_]['new_mw']/df_.loc[idx_]['spec_mw'])

idx_ = df_['scenario'] == 'VREhigh_SThigh_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid'
print(df_.loc[idx_, 'new_mw']/df_.loc[idx_, 'spec_mw'])
```

State-level imports and exports depend on the differences between the state's electricity demand and generation potential but also the state's location in the overall network where it could be a conduit for electricity transfers across regions. Rajasthan is the largest net exporter of energy because of its large solar and wind deployment relative to its demand in 2050. The state's exports, xx\% of its own in-state electricity demand, help meet demand in capacity-deficit northern states including Delhi and Haryana. Most other large western and southern states with the exception of Kerala and Telangana are net exporters (Karnataka, Madhya Pradesh, Tamil Nadu, and Gujarat) or close to zero net electricity exchanges (Andhra Pradesh and Maharashtra). Most large northern and eastern states including Delhi, Haryana, West Bengal, Uttar Pradesh, Odisha, Bihar, and Punjab) are net importers, mainly because of their relatively poor renewable resources. At the same time, many states experience substantial imports and exports, much larger than their net transfers, highlighting the importance of transmission and electricity transfers to balance the diurnal and seasonal variability of solar and wind across India.

## Cost-optimal deployment of renewable energy and battery storage in 2040 is three times the official targets in 2030

Even after imposing a clean energy limit in the near term (2030) or a carbon target of 80-100% in the long term (2050), the costs in the medium term (2040) are similar to a scenario in which no clean energy or carbon targets are applied. This result shows that with medium cost decline trajectories of VRE and storage, a conservative assumption of low coal costs, and a climate change-impacted and solar-aligned future electricity demand, carbon emissions can reduce 20.4\% (32-11.9\% for high and low VRE and storage costs) by 2040. 

```python
df_ = pd.read_csv(path_to_tables + '/cost-emissions.csv')
print(df_['scenario'].unique())

df_1_ = df_.loc[df_['scenario'] == 'VRElow_STlow_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid']
df_2_ = df_.loc[df_['scenario'] == 'VREmid_STmid_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid']
df_3_ = df_.loc[df_['scenario'] == 'VREhigh_SThigh_CONVmid_H2_RES_8PRM_CC_50RPS_90CAP_500GW_PIERmid']

df_1_ = df_1_.set_index(["period"])
df_2_ = df_2_.set_index(["period"])
df_3_ = df_3_.set_index(["period"])

print(1. - df_1_.loc[2040, 'carbon_emissions_tons']/df_1_.loc[2020, 'carbon_emissions_tons'])
print(1. - df_2_.loc[2040, 'carbon_emissions_tons']/df_2_.loc[2020, 'carbon_emissions_tons'])
print(1. - df_3_.loc[2040, 'carbon_emissions_tons']/df_3_.loc[2020, 'carbon_emissions_tons'])
```

Under this scenario, solar and wind capacity in 2040 is 995.2GW and 433.5GW, respectively, generating 70\% of electricity, and battery power and energy storage capacities are 400.8GW and 801.6GWh, capacities which are approximately three times the 2030 official targets. These results could inform India's next set of clean energy deployment targets.
