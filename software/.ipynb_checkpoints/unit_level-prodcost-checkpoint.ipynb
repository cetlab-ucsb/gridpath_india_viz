```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

path_to_scenario = '/Users/Guille/Desktop/india_power/scenarios/'
path_to_images   = '/Users/Guille/Desktop/india_power/images/'
```

```python
def _load_scen(scenario, path_to_scenario):
    return pd.read_csv(path_to_scenario + scenario + f'/results/project_timepoint.csv', 
                       low_memory = False)

scenario = 'prodcost_2050_1prob_circular_VREmid_STmid_CONVmid_CC_RES_H2_10PRM_50RPS_90CAP_PIERmid'

df_ = _load_scen(scenario, path_to_scenario)
print(df_.shape)
```

```python
for project in df_.loc[((df_['operational_type'] == 'stor')
                        & (df_['balancing_type'] == 'day')), 'project'].unique()[:1]:

    soc_ = df_.loc[df_['project'] == project, 'starting_energy_mwh'].to_numpy()
    time_ = df_.loc[df_['project'] == project, 'timepoint'].to_numpy()


    plt.figure(figsize = (15, 2.5))
    plt.title(project)
    plt.plot(soc_.reshape(-1, 24).T)
    
    print((soc_.reshape(-1, 24)[:, 0] < soc_.reshape(-1, 24)[:, -1]).sum())

    plt.savefig(path_to_images + f'{project}_{scenario}.png', 
                bbox_inches = 'tight', 
                dpi         = 100)
    plt.show()
```

```python
technologies_ = ['Subcritical_Coal_Large',
                 'Subcritical_Coal_Small',
                 'Supercritical_Coal',
                 'CCGT',
                 'Diesel']

sectors_ = ['private',
            'state',
            'central',
            'pipeline',
            'new']

df_p_ = df_[['project', 
             'period', 
             'technology', 
             'load_zone',
             'number_of_hours_in_timepoint',
             'timepoint_weight',
             'capacity_mw',
             'power_mw']].copy()

print(df_p_.shape)
print(df_p_['technology'].unique())

df_p_ = df_p_.loc[df_p_['technology'].isin(technologies_)].reset_index(drop = True)
print(df_p_.shape)

df_p_['power_mw'] = df_p_['power_mw'] * df_p_['number_of_hours_in_timepoint'] * df_p_['timepoint_weight'] 
print(df_p_.shape)

df_p_ = df_p_.groupby(['project', 
                       'period', 
                       'load_zone',
                       'technology', 
                       'capacity_mw']).agg({'power_mw': 'sum'}).reset_index(drop = False)
print(df_p_.shape)
df_p_['sector'] = ''
for sector in sectors_:
    idx_ = df_p_.project.str.contains(sector)

    df_p_.loc[idx_, 'sector'] = sector
    print(idx_.sum(), sector)

print(df_p_)
```
